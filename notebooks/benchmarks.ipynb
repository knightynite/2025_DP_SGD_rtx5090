# 2025 — DP-SGD on RTX 5090 — Benchmarks & Tradeoffs

From-scratch DP-SGD with per-sample clipping + Gaussian noise. Throughput on Blackwell (RTX 5090, CUDA 13.1) vs Hopper (H100). Privacy/accuracy tradeoff for MNIST and CIFAR-10.

```python
import json, matplotlib.pyplot as plt
%matplotlib inline
```

## Privacy budget growth

```python
from IPython.display import Image
Image('figures/privacy_curve.png')
```

ε grows roughly as √steps — the standard DP-SGD shape. The RDP accountant gives tight bounds; the moments accountant would track higher.

## Privacy / accuracy tradeoff

```python
Image('figures/privacy_accuracy_tradeoff.png')
```

MNIST is forgiving — even at ε=1, accuracy is >93%. CIFAR-10 is brutal — ε=1 collapses to ~32%, and you need ε≥8 to recover useful performance. This matches the literature.

## Throughput: H100 vs RTX 5090

```python
Image('figures/throughput_comparison.png')
```

Blackwell roughly 2× H100 in BF16 throughput, ~3× in FP8. Per-sample-grad DP-SGD becomes practical on consumer hardware for the first time.
